**Part 1: Setup and Import Libraries**

```python
# Part 1: Import all necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Scikit-learn core
from sklearn.model_selection import train_test_split, StratifiedKFold, RandomizedSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer          # <-- fixed import
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# Metrics
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, confusion_matrix, precision_recall_curve,
                             auc, brier_score_loss, ConfusionMatrixDisplay)

# Classifiers
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from xgboost import XGBClassifier

# Imbalanced learning
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline

# Feature selection and SHAP
from sklearn.feature_selection import RFECV
import shap

# Set random seed
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

print("All libraries imported successfully.")
```

**Part 2: Load and Explore Data**

```python
# Part 2: Load the Excel file and perform initial exploration

df = pd.read_excel(r"C:\Users\HP\Desktop\research word documents\viral load\updated_research_nutrition_data.xlsx")

print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())
print("\nData types:")
print(df.dtypes)
print("\nTarget variable distribution (TPT Started):")
print(df["TPT Started"].value_counts())
```

**Part 3: Data Preprocessing**

```python
# 3.1 Separate features and target
target_col = "TPT Started"
X = df.drop(columns=[target_col])
y = df[target_col].map({"Yes": 1, "No": 0})   # Binary encoding

print("Features shape:", X.shape)
print("Target distribution:\n", y.value_counts())
```

```python
# 3.2 Identify numeric and categorical columns
numeric_cols = X.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = X.select_dtypes(include=["object"]).columns.tolist()

print("Numeric columns:", numeric_cols)
print("Categorical columns:", categorical_cols)
```

```python
# 3.3 Replace zeros with NaN for CD4 columns
for col in ["Baseline CD4 Count", "Recent CD4 Count"]:
    if col in numeric_cols:
        X[col] = X[col].replace(0, np.nan)
        print(f"Replaced zeros with NaN in {col}: {X[col].isna().sum()} missing values")
```

```python
# 3.4 Impute numeric missing with median
numeric_imputer = SimpleImputer(strategy="median")
X_numeric_imputed = numeric_imputer.fit_transform(X[numeric_cols])
X_numeric_imputed = pd.DataFrame(X_numeric_imputed, columns=numeric_cols, index=X.index)

print("Numeric imputation complete. Missing values after imputation:", 
      pd.DataFrame(X_numeric_imputed).isna().sum().sum())
```

```python
# 3.5 Impute categorical missing with most frequent
categorical_imputer = SimpleImputer(strategy="most_frequent")
X_categorical_imputed = categorical_imputer.fit_transform(X[categorical_cols])
X_categorical_imputed = pd.DataFrame(X_categorical_imputed, columns=categorical_cols, index=X.index)

print("Categorical imputation complete.")
```

```python
# 3.6 Combine numeric and categorical
X_processed = pd.concat([X_numeric_imputed, X_categorical_imputed], axis=1)

# One-hot encode categorical columns (drop first to avoid multicollinearity)
X_encoded = pd.get_dummies(X_processed, columns=categorical_cols, drop_first=True)

print("Shape after one-hot encoding:", X_encoded.shape)
print("Column names (first 20):", X_encoded.columns[:20].tolist())
```

```python
# Part 4: Split data into training and testing sets (stratified)
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)

print(f"Training set: {X_train.shape[0]} samples")
print(f"Test set: {X_test.shape[0]} samples")
print("Training TPT distribution:\n", y_train.value_counts(normalize=True))
print("Test TPT distribution:\n", y_test.value_counts(normalize=True))

# Note: Scaling will be applied inside pipelines for models that need it.
```

```python
# Part 5: Dictionary of models and their hyperparameter search spaces
models = {
    "Logistic Regression": {
        "model": LogisticRegression(random_state=RANDOM_STATE, max_iter=1000),
        "params": {
            "classifier__C": [0.01, 0.1, 1, 10, 100],
            "classifier__penalty": ["l1", "l2"],
            "classifier__solver": ["liblinear"]
        },
        "scale": True
    },
    "Decision Tree": {
        "model": DecisionTreeClassifier(random_state=RANDOM_STATE),
        "params": {
            "classifier__max_depth": [3, 5, 10, 15, None],
            "classifier__min_samples_split": [2, 5, 10],
            "classifier__min_samples_leaf": [1, 2, 4]
        },
        "scale": False
    },
    "Random Forest": {
        "model": RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1),
        "params": {
            "classifier__n_estimators": [100, 200, 300],
            "classifier__max_depth": [10, 20, 30, None],
            "classifier__min_samples_split": [2, 5, 10],
            "classifier__min_samples_leaf": [1, 2, 4],
            "classifier__max_features": ["sqrt", "log2"]
        },
        "scale": False
    },
    "XGBoost": {
        "model": XGBClassifier(random_state=RANDOM_STATE, eval_metric="logloss", use_label_encoder=False),
        "params": {
            "classifier__n_estimators": [100, 200, 300],
            "classifier__max_depth": [3, 6, 10],
            "classifier__learning_rate": [0.01, 0.05, 0.1],
            "classifier__subsample": [0.7, 0.8, 1.0]
        },
        "scale": False
    },
    "SVM": {
        "model": SVC(random_state=RANDOM_STATE, probability=True),
        "params": {
            "classifier__C": [0.1, 1, 10, 100],
            "classifier__gamma": ["scale", "auto"],
            "classifier__kernel": ["rbf"]
        },
        "scale": True
    },
    "k-NN": {
        "model": KNeighborsClassifier(),
        "params": {
            "classifier__n_neighbors": [3, 5, 7, 11, 15],
            "classifier__weights": ["uniform", "distance"],
            "classifier__metric": ["euclidean", "manhattan"]
        },
        "scale": True
    }
}

print("Models and hyperparameter grids defined.")
```

```python
# Part 6: Loop over models, apply SMOTE, tune, and evaluate on test set
results = []
best_estimators = {}
feature_importances = {}

for name, config in models.items():
    print(f"\n{'='*60}\nTraining {name}\n{'='*60}")
    
    # Build pipeline: scaler (if needed) -> SMOTE -> classifier
    steps = []
    if config["scale"]:
        steps.append(("scaler", StandardScaler()))
    steps.append(("smote", SMOTE(random_state=RANDOM_STATE)))
    steps.append(("classifier", config["model"]))
    
    pipeline = ImbPipeline(steps)
    
    # Use RandomizedSearchCV for hyperparameter tuning
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
    search = RandomizedSearchCV(
        pipeline, config["params"], n_iter=20, cv=cv, scoring="f1",
        random_state=RANDOM_STATE, n_jobs=-1, verbose=1
    )
    search.fit(X_train, y_train)
    
    best_model = search.best_estimator_
    best_estimators[name] = best_model
    print(f"Best parameters: {search.best_params_}")
    print(f"Best cross-validated F1: {search.best_score_:.4f}")
    
    # Predict on test set
    y_pred = best_model.predict(X_test)
    y_proba = best_model.predict_proba(X_test)[:, 1] if hasattr(best_model, "predict_proba") else None
    
    # Calculate metrics
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred)
    rec = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)
    auc_roc = roc_auc_score(y_test, y_proba) if y_proba is not None else np.nan
    
    if y_proba is not None:
        prec_vals, rec_vals, _ = precision_recall_curve(y_test, y_proba)
        auc_pr = auc(rec_vals, prec_vals)
        brier = brier_score_loss(y_test, y_proba)
    else:
        auc_pr = np.nan
        brier = np.nan
    
    results.append({
        "Model": name,
        "Accuracy": acc,
        "Precision": prec,
        "Recall": rec,
        "F1": f1,
        "AUC-ROC": auc_roc,
        "AUC-PR": auc_pr,
        "Brier Score": brier
    })
    
    print(f"Test Performance: Acc={acc:.4f}, Prec={prec:.4f}, Rec={rec:.4f}, F1={f1:.4f}, AUC-ROC={auc_roc:.4f}")
    
    # Store feature importances for tree-based models
    if name in ["Random Forest", "Decision Tree", "XGBoost"]:
        if hasattr(best_model.named_steps["classifier"], "feature_importances_"):
            importances = best_model.named_steps["classifier"].feature_importances_
            feature_importances[name] = pd.Series(importances, index=X_encoded.columns)
```

```python
# Part 7: Create results table and identify best model
results_df = pd.DataFrame(results)
print("\n\n=== Model Performance Summary ===")
print(results_df.round(4))

# Save to CSV
results_df.to_csv("model_performance.csv", index=False)
print("Performance table saved as 'model_performance.csv'.")

# Best model by F1 and by AUC-ROC
best_f1_model = results_df.loc[results_df["F1"].idxmax(), "Model"]
best_auc_model = results_df.loc[results_df["AUC-ROC"].idxmax(), "Model"]
print(f"\nBest model by F1: {best_f1_model} (F1={results_df['F1'].max():.4f})")
print(f"Best model by AUC-ROC: {best_auc_model} (AUC={results_df['AUC-ROC'].max():.4f})")
```

```python
# 8.1 Plot top 10 features from Random Forest (if available)
rf_importance = feature_importances.get("Random Forest", None)
if rf_importance is not None:
    rf_importance_sorted = rf_importance.sort_values(ascending=False).head(10)
    plt.figure(figsize=(10,6))
    sns.barplot(x=rf_importance_sorted.values, y=rf_importance_sorted.index)
    plt.title("Top 10 Features - Random Forest Importance")
    plt.xlabel("Mean Decrease in Impurity")
    plt.tight_layout()
    plt.savefig("rf_feature_importance.png")
    plt.show()
else:
    print("Random Forest importance not available.")
```

```python
# 8.2 RFECV to select most predictive features
print("\nRunning Recursive Feature Elimination with Random Forest (RFECV)...")
# Get the classifier from the best Random Forest pipeline (or train a new one)
if "Random Forest" in best_estimators:
    rf_clf = best_estimators["Random Forest"].named_steps["classifier"]
else:
    rf_clf = RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1)

# Pipeline with SMOTE + RF
rfecv_pipeline = ImbPipeline([
    ("smote", SMOTE(random_state=RANDOM_STATE)),
    ("rf", rf_clf)
])

rfecv = RFECV(
    estimator=rfecv_pipeline, step=1, cv=StratifiedKFold(5, shuffle=True, random_state=RANDOM_STATE),
    scoring="f1", n_jobs=-1, verbose=1
)
rfecv.fit(X_train, y_train)

selected_features = X_encoded.columns[rfecv.support_]
print(f"Number of features selected by RFECV: {len(selected_features)}")
print("Selected features (first 20):", list(selected_features)[:20])

# Feature ranking
feature_ranking = pd.Series(rfecv.ranking_, index=X_encoded.columns).sort_values()
print("\nTop 10 features by RFECV ranking:")
print(feature_ranking.head(10))
```

```python
# 8.3 SHAP analysis on the best performing model (by AUC-ROC)
best_model_name = best_auc_model
best_model_obj = best_estimators[best_model_name]
print(f"\nPerforming SHAP analysis on best model: {best_model_name}")

# Use a sample of test data for speed
X_test_sample = X_test.sample(n=min(1000, len(X_test)), random_state=RANDOM_STATE)

if best_model_name in ["Random Forest", "Decision Tree", "XGBoost"]:
    clf = best_model_obj.named_steps["classifier"]
    explainer = shap.TreeExplainer(clf)
    shap_values = explainer.shap_values(X_test_sample)
    # For binary classification, shap_values is a list of two arrays; use class 1
    if isinstance(shap_values, list):
        shap_values = shap_values[1]
    
    # Summary plot
    shap.summary_plot(shap_values, X_test_sample, feature_names=X_encoded.columns, show=False)
    plt.savefig("shap_summary_plot.png", bbox_inches="tight")
    plt.show()
    
    # Mean absolute SHAP values
    mean_shap = np.abs(shap_values).mean(axis=0)
    shap_importance = pd.Series(mean_shap, index=X_encoded.columns).sort_values(ascending=False)
    print("\nTop 10 features by SHAP importance:")
    print(shap_importance.head(10))
    shap_importance.to_csv("shap_importance.csv")
else:
    print(f"SHAP TreeExplainer not applicable for {best_model_name}. Skipping SHAP (or use KernelExplainer for advanced users).")
```

```python
# Part 9: Confusion matrix for the best model
y_pred_best = best_estimators[best_auc_model].predict(X_test)
cm = confusion_matrix(y_test, y_pred_best)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["No TPT", "TPT"])
disp.plot()
plt.title(f"Confusion Matrix - {best_auc_model}")
plt.savefig("confusion_matrix_best_model.png")
plt.show()

print("\nAll outputs have been saved:")
print(" - model_performance.csv")
print(" - rf_feature_importance.png (if Random Forest was trained)")
print(" - shap_importance.csv (if best model was tree-based)")
print(" - shap_summary_plot.png")
print(" - confusion_matrix_best_model.png")
```

```python
# 8.4 Waterfall plots for individual predictions (requires tree-based model for best performance)
# This code will work with TreeExplainer (Random Forest, XGBoost, Decision Tree)
# For non-tree models, it will attempt KernelExplainer (may be slow) or display a message.

best_model_name = best_auc_model
best_model_obj = best_estimators[best_model_name]

tree_based_models = ["Random Forest", "Decision Tree", "XGBoost"]
is_tree_based = best_model_name in tree_based_models

if not is_tree_based:
    print(f"Best model ({best_model_name}) is not tree-based. Waterfall plots will use KernelExplainer (slower).")
    print("Consider re-running with a tree-based model for faster waterfall plots.")
else:
    print(f"Using TreeExplainer for {best_model_name}.")

# Select a few interesting test samples:
# - A correct prediction where true label = 1 (TPT user)
# - A correct prediction where true label = 0 (non-TPT user)
# - A misclassified sample (if any)

y_pred_best = best_model_obj.predict(X_test)
y_proba_best = best_model_obj.predict_proba(X_test)[:, 1]

# Get indices of correct TPT, correct non-TPT, and misclassified
correct_tpt = (y_test == 1) & (y_pred_best == 1)
correct_no_tpt = (y_test == 0) & (y_pred_best == 0)
misclassified = (y_test != y_pred_best)

# Choose one sample from each category (if available)
sample_indices = []
if correct_tpt.any():
    sample_indices.append(correct_tpt[correct_tpt].index[0])
if correct_no_tpt.any():
    sample_indices.append(correct_no_tpt[correct_no_tpt].index[0])
if misclassified.any():
    sample_indices.append(misclassified[misclassified].index[0])

# If no misclassified, add another correct TPT or correct non-TPT
if len(sample_indices) < 2 and correct_tpt.any():
    sample_indices.append(correct_tpt[correct_tpt].index[1] if correct_tpt.sum() > 1 else correct_tpt[correct_tpt].index[0])
if len(sample_indices) < 3 and correct_no_tpt.any():
    sample_indices.append(correct_no_tpt[correct_no_tpt].index[1] if correct_no_tpt.sum() > 1 else correct_no_tpt[correct_no_tpt].index[0])

print(f"Generating waterfall plots for {len(sample_indices)} samples.")

# For tree-based models: use TreeExplainer
if is_tree_based:
    clf = best_model_obj.named_steps["classifier"]
    explainer = shap.TreeExplainer(clf)
    # Get SHAP values for the selected samples
    X_sample = X_test.loc[sample_indices]
    shap_values_sample = explainer.shap_values(X_sample)
    # For binary classification, shap_values_sample may be a list; use class 1
    if isinstance(shap_values_sample, list):
        shap_values_sample = shap_values_sample[1]
    
    # Base value (expected model output)
    base_value = explainer.expected_value
    if isinstance(base_value, list):
        base_value = base_value[1]
    
    # Plot waterfall for each sample
    for i, idx in enumerate(sample_indices):
        plt.figure(figsize=(12, 6))
        shap.plots.waterfall(
            shap.Explanation(
                values=shap_values_sample[i],
                base_values=base_value,
                data=X_sample.iloc[i],
                feature_names=X_encoded.columns.tolist()
            ),
            show=False,
            max_display=10
        )
        true_label = y_test[idx]
        pred_label = y_pred_best[idx]
        proba = y_proba_best[X_test.index.get_loc(idx)]
        plt.title(f"Sample {idx} | True: {true_label}, Predicted: {pred_label} | Probability: {proba:.3f}")
        plt.tight_layout()
        plt.savefig(f"waterfall_sample_{idx}.png", bbox_inches="tight", dpi=300)
        plt.show()
        print(f"Waterfall plot saved as waterfall_sample_{idx}.png")

# For non-tree models: use KernelExplainer (slower, but possible)
else:
    # Use a smaller background and fewer samples to keep runtime reasonable
    background = X_train.sample(n=min(100, len(X_train)), random_state=RANDOM_STATE)
    # Define prediction function
    def predict_proba_wrapper(x):
        return best_model_obj.predict_proba(x)[:, 1]
    
    print("Initializing KernelExplainer (this may take a minute)...")
    explainer = shap.KernelExplainer(predict_proba_wrapper, background)
    # Compute SHAP values for selected samples
    X_sample = X_test.loc[sample_indices]
    shap_values_sample = explainer.shap_values(X_sample, nsamples=200)  # nsamples = number of permutations
    base_value = explainer.expected_value
    if isinstance(base_value, list):
        base_value = base_value[1]
    
    for i, idx in enumerate(sample_indices):
        plt.figure(figsize=(12, 6))
        shap.plots.waterfall(
            shap.Explanation(
                values=shap_values_sample[i],
                base_values=base_value,
                data=X_sample.iloc[i],
                feature_names=X_encoded.columns.tolist()
            ),
            show=False,
            max_display=10
        )
        true_label = y_test[idx]
        pred_label = y_pred_best[idx]
        proba = y_proba_best[X_test.index.get_loc(idx)]
        plt.title(f"Sample {idx} | True: {true_label}, Predicted: {pred_label} | Probability: {proba:.3f}")
        plt.tight_layout()
        plt.savefig(f"waterfall_sample_{idx}.png", bbox_inches="tight", dpi=300)
        plt.show()
        print(f"Waterfall plot saved as waterfall_sample_{idx}.png")
```
